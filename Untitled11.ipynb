```python
# ============================================================
# ECC5121 Assignment I
# Topic 5: Massive MIMO Antenna Panel Deactivation
# Final Optimized Google Colab Simulation Code
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os
import zipfile
from google.colab import files

# ============================================================
# 1. Reproducibility and Output Folder
# ============================================================

np.random.seed(42)

output_dir = "Topic5_Massive_MIMO_Final_Optimized_Outputs"
os.makedirs(output_dir, exist_ok=True)

# ============================================================
# 2. System Configuration
# ============================================================

# Massive MIMO array configuration
total_panels = 8
antennas_per_panel = 8
total_antennas = total_panels * antennas_per_panel   # 64 antennas

# Main simulation parameters
cell_radius = 500                      # meters
bandwidth_hz = 20e6                   # 20 MHz
tx_power_dbm = 43                     # dBm
tx_power_watt = 10 ** ((tx_power_dbm - 30) / 10)

noise_power_dbm = -94
noise_power_watt = 10 ** ((noise_power_dbm - 30) / 10)

# Final QoS threshold used in the main simulation
sinr_threshold_db = 2.4
sinr_threshold_linear = 10 ** (sinr_threshold_db / 10)

# Monte Carlo trials
num_trials = 300

# Representative traffic load scenarios
traffic_scenarios = {
    "Low Load": 5,
    "Medium Load": 15,
    "Peak Load": 30
}

# Panel activation strategies
strategies = {
    "Baseline Always-On": {
        "Low Load": 8,
        "Medium Load": 8,
        "Peak Load": 8
    },
    "Balanced Adaptive": {
        "Low Load": 4,
        "Medium Load": 6,
        "Peak Load": 8
    },
    "Aggressive Adaptive": {
        "Low Load": 2,
        "Medium Load": 4,
        "Peak Load": 8
    }
}

load_order = ["Low Load", "Medium Load", "Peak Load"]
strategy_order = ["Baseline Always-On", "Balanced Adaptive", "Aggressive Adaptive"]

# ============================================================
# 3. Synthetic Diurnal Traffic Input
#    This addresses the assignment requirement for synthetic traffic patterns.
# ============================================================

hours = np.arange(24)

# Synthetic expected daily traffic profile:
# low at night, moderate in daytime, peak in the evening
expected_users = (
    5
    + 8 * np.exp(-((hours - 12) ** 2) / 22)
    + 20 * np.exp(-((hours - 20) ** 2) / 14)
)

# Poisson-generated synthetic user counts
synthetic_user_counts = np.random.poisson(expected_users)

plt.figure(figsize=(8, 5))
plt.plot(hours, synthetic_user_counts, marker="o", linewidth=2, label="Synthetic user count")
plt.plot(hours, expected_users, linestyle="--", linewidth=2, label="Expected traffic intensity")
plt.xlabel("Hour of Day")
plt.ylabel("Number of Active Users")
plt.title("Synthetic Diurnal Traffic Profile")
plt.xticks(hours)
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.savefig(os.path.join(output_dir, "Figure_0_Synthetic_Diurnal_Traffic_Profile.png"), dpi=300)
plt.show()

# ============================================================
# 4. Component-Based Base Station Power Model
# ============================================================

P_static = 300
P_rf_chain = 4
P_baseband_coeff = 0.08
P_cooling_ratio = 0.15

def calculate_power(active_antennas, num_users):
    """
    Component-based BS power model:
    P_total = (P_static + P_RF + P_BB) * (1 + eta)
    """
    rf_power = active_antennas * P_rf_chain
    baseband_power = P_baseband_coeff * active_antennas * num_users
    subtotal = P_static + rf_power + baseband_power
    cooling_power = P_cooling_ratio * subtotal
    total_power = subtotal + cooling_power
    return total_power

# ============================================================
# 5. User Spatial Distribution and Channel Model
# ============================================================

def generate_user_positions(num_users, radius):
    """
    Uniform user distribution in a circular cell.
    theta is uniformly distributed over [0, 2π), creating angular spread.
    """
    r = radius * np.sqrt(np.random.rand(num_users))
    theta = 2 * np.pi * np.random.rand(num_users)

    x = r * np.cos(theta)
    y = r * np.sin(theta)

    distances = np.sqrt(x**2 + y**2)
    distances = np.maximum(distances, 10)

    return distances, theta

def pathloss_model(distances):
    """
    Simplified urban macro path loss model:
    PL(dB) = 128.1 + 37.6 log10(d_km)
    """
    distances_km = distances / 1000
    pathloss_db = 128.1 + 37.6 * np.log10(distances_km)
    pathloss_linear = 10 ** (-pathloss_db / 10)
    return pathloss_linear

def generate_full_channel(num_users, total_antennas, pathloss_linear):
    """
    Rayleigh fading channel with large-scale path loss.
    H shape: users x antennas
    """
    small_scale = (
        np.random.randn(num_users, total_antennas)
        + 1j * np.random.randn(num_users, total_antennas)
    ) / np.sqrt(2)

    H = small_scale * np.sqrt(pathloss_linear[:, np.newaxis])
    return H

# ============================================================
# 6. MRT Beamforming and QoS Calculation
# ============================================================

def mrt_beamforming(H):
    """
    Maximum Ratio Transmission beamforming.
    """
    W = np.conj(H)
    norms = np.linalg.norm(W, axis=1, keepdims=True)
    W = W / (norms + 1e-12)
    return W

def calculate_sinr(H, W, tx_power, noise_power):
    """
    Downlink SINR calculation for each active user.
    """
    num_users = H.shape[0]
    power_per_user = tx_power / num_users
    sinr_values = []

    for k in range(num_users):
        hk = H[k, :]
        desired_signal = power_per_user * np.abs(np.dot(hk, W[k, :])) ** 2

        interference = 0
        for j in range(num_users):
            if j != k:
                interference += power_per_user * np.abs(np.dot(hk, W[j, :])) ** 2

        sinr = desired_signal / (interference + noise_power)
        sinr_values.append(sinr)

    return np.array(sinr_values)

# ============================================================
# 7. Main Monte Carlo Simulation
# ============================================================

results_storage = {}

for load_name in load_order:
    num_users = traffic_scenarios[load_name]

    for strategy_name in strategy_order:
        results_storage[(load_name, strategy_name)] = {
            "power": [],
            "avg_sinr_db": [],
            "avg_throughput_mbps": [],
            "coverage": []
        }

    for trial in range(num_trials):

        # Same user/channel realization reused across strategies for fair comparison
        distances, theta = generate_user_positions(num_users, cell_radius)
        pathloss = pathloss_model(distances)
        H_full = generate_full_channel(num_users, total_antennas, pathloss)

        for strategy_name in strategy_order:
            active_panels = strategies[strategy_name][load_name]
            active_antennas = active_panels * antennas_per_panel

            H = H_full[:, :active_antennas]
            W = mrt_beamforming(H)
            sinr = calculate_sinr(H, W, tx_power_watt, noise_power_watt)

            avg_sinr_db = 10 * np.log10(np.mean(sinr))
            throughput = bandwidth_hz * np.log2(1 + sinr)
            avg_throughput_mbps = np.mean(throughput) / 1e6
            coverage = np.mean(sinr >= sinr_threshold_linear)
            power = calculate_power(active_antennas, num_users)

            results_storage[(load_name, strategy_name)]["power"].append(power)
            results_storage[(load_name, strategy_name)]["avg_sinr_db"].append(avg_sinr_db)
            results_storage[(load_name, strategy_name)]["avg_throughput_mbps"].append(avg_throughput_mbps)
            results_storage[(load_name, strategy_name)]["coverage"].append(coverage)

# Convert simulation results to DataFrame
results = []

for load_name in load_order:
    num_users = traffic_scenarios[load_name]

    for strategy_name in strategy_order:
        active_panels = strategies[strategy_name][load_name]
        active_antennas = active_panels * antennas_per_panel
        stored = results_storage[(load_name, strategy_name)]

        results.append({
            "Load Scenario": load_name,
            "Strategy": strategy_name,
            "Users": num_users,
            "Active Panels": active_panels,
            "Active Antennas": active_antennas,
            "Power Consumption (W)": np.mean(stored["power"]),
            "Average SINR (dB)": np.mean(stored["avg_sinr_db"]),
            "Average Throughput (Mbps)": np.mean(stored["avg_throughput_mbps"]),
            "Coverage Probability": np.mean(stored["coverage"])
        })

df = pd.DataFrame(results)

# Calculate energy saving relative to the baseline
baseline_power = df[df["Strategy"] == "Baseline Always-On"][
    ["Load Scenario", "Power Consumption (W)"]
].rename(columns={"Power Consumption (W)": "Baseline Power (W)"})

df = df.merge(baseline_power, on="Load Scenario")
df["Energy Saving (%)"] = (
    (df["Baseline Power (W)"] - df["Power Consumption (W)"])
    / df["Baseline Power (W)"]
    * 100
)

df["Load Scenario"] = pd.Categorical(df["Load Scenario"], categories=load_order, ordered=True)
df["Strategy"] = pd.Categorical(df["Strategy"], categories=strategy_order, ordered=True)
df = df.sort_values(["Load Scenario", "Strategy"])

# Save main numerical results
main_results_path = os.path.join(output_dir, "simulation_results_final.csv")
df.to_csv(main_results_path, index=False)

display(df.round(3))

# ============================================================
# 8. Plotting Function
# ============================================================

def save_line_plot(y_column, y_label, title, filename, ylim=None, coverage_line=False):
    plt.figure(figsize=(8, 5))

    for strategy in strategy_order:
        subset = df[df["Strategy"] == strategy]
        plt.plot(
            subset["Load Scenario"].astype(str),
            subset[y_column],
            marker="o",
            linewidth=2,
            label=strategy
        )

    if coverage_line:
        plt.axhline(
            y=0.9,
            linestyle="--",
            linewidth=2,
            label="90% Coverage Requirement"
        )

    plt.xlabel("Traffic Load Scenario")
    plt.ylabel(y_label)
    plt.title(title)
    plt.legend()
    plt.grid(True)

    if ylim is not None:
        plt.ylim(ylim)

    plt.tight_layout()
    plt.savefig(os.path.join(output_dir, filename), dpi=300)
    plt.show()

# ============================================================
# 9. Core Figures
# ============================================================

save_line_plot(
    "Power Consumption (W)",
    "Power Consumption (W)",
    "Power Consumption under Different Traffic Load Scenarios",
    "Figure_1_Power_Consumption.png"
)

save_line_plot(
    "Energy Saving (%)",
    "Energy Saving (%)",
    "Energy Saving Compared with Always-On Baseline",
    "Figure_2_Energy_Saving.png"
)

save_line_plot(
    "Average SINR (dB)",
    "Average SINR (dB)",
    "Average SINR under Different Strategies",
    "Figure_3_Average_SINR.png"
)

save_line_plot(
    "Average Throughput (Mbps)",
    "Average Throughput (Mbps)",
    "Average Throughput under Different Strategies",
    "Figure_4_Average_Throughput.png"
)

save_line_plot(
    "Coverage Probability",
    "Coverage Probability",
    "Coverage Probability under Different Strategies",
    "Figure_5_Coverage_Probability.png",
    ylim=(0, 1.05),
    coverage_line=True
)

# Figure 6: Energy saving vs coverage trade-off
plt.figure(figsize=(8, 5))

label_positions = {
    ("Baseline Always-On", "Low Load"): (0.4, 0.015),
    ("Baseline Always-On", "Medium Load"): (0.4, -0.025),
    ("Baseline Always-On", "Peak Load"): (0.4, 0.015),

    ("Balanced Adaptive", "Low Load"): (0.4, 0.015),
    ("Balanced Adaptive", "Medium Load"): (0.4, -0.025),
    ("Balanced Adaptive", "Peak Load"): (0.4, 0.015),

    ("Aggressive Adaptive", "Low Load"): (0.4, 0.015),
    ("Aggressive Adaptive", "Medium Load"): (0.4, -0.025),
    ("Aggressive Adaptive", "Peak Load"): (0.4, 0.015),
}

for strategy in strategy_order:
    subset = df[df["Strategy"] == strategy]

    plt.scatter(
        subset["Energy Saving (%)"],
        subset["Coverage Probability"],
        s=90,
        label=strategy
    )

    for _, row in subset.iterrows():
        key = (str(row["Strategy"]), str(row["Load Scenario"]))
        dx, dy = label_positions[key]
        plt.text(
            row["Energy Saving (%)"] + dx,
            row["Coverage Probability"] + dy,
            str(row["Load Scenario"]),
            fontsize=8
        )

plt.axhline(y=0.9, linestyle="--", linewidth=2, label="90% Coverage Requirement")
plt.xlabel("Energy Saving (%)")
plt.ylabel("Coverage Probability")
plt.title("Energy Saving vs Coverage Trade-off")
plt.ylim(0, 1.05)
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.savefig(os.path.join(output_dir, "Figure_6_Energy_Saving_vs_Coverage.png"), dpi=300)
plt.show()

# ============================================================
# 10. Sensitivity Test: Coverage Probability vs SINR Threshold
#     This is stronger and more meaningful than the previous radius test.
# ============================================================

sensitivity_thresholds_db = [2.0, 2.4, 3.0, 3.5, 4.0]
sensitivity_load = "Medium Load"
sensitivity_users = traffic_scenarios[sensitivity_load]
sensitivity_trials = 300

sensitivity_results = []

for threshold_db in sensitivity_thresholds_db:
    threshold_linear = 10 ** (threshold_db / 10)

    for strategy_name in strategy_order:
        coverages = []

        for trial in range(sensitivity_trials):
            distances, theta = generate_user_positions(sensitivity_users, cell_radius)
            pathloss = pathloss_model(distances)
            H_full = generate_full_channel(sensitivity_users, total_antennas, pathloss)

            active_panels = strategies[strategy_name][sensitivity_load]
            active_antennas = active_panels * antennas_per_panel

            H = H_full[:, :active_antennas]
            W = mrt_beamforming(H)
            sinr = calculate_sinr(H, W, tx_power_watt, noise_power_watt)

            coverages.append(np.mean(sinr >= threshold_linear))

        sensitivity_results.append({
            "SINR Threshold (dB)": threshold_db,
            "Strategy": strategy_name,
            "Coverage Probability": np.mean(coverages)
        })

sensitivity_df = pd.DataFrame(sensitivity_results)
sensitivity_df.to_csv(
    os.path.join(output_dir, "sensitivity_results_coverage_vs_sinr_threshold.csv"),
    index=False
)

plt.figure(figsize=(8, 5))

for strategy in strategy_order:
    subset = sensitivity_df[sensitivity_df["Strategy"] == strategy]
    plt.plot(
        subset["SINR Threshold (dB)"],
        subset["Coverage Probability"],
        marker="o",
        linewidth=2,
        label=strategy
    )

plt.axhline(y=0.9, linestyle="--", linewidth=2, label="90% Coverage Requirement")
plt.xlabel("SINR Threshold (dB)")
plt.ylabel("Coverage Probability")
plt.title("Sensitivity of Coverage Probability to SINR Threshold\n(Medium Load Scenario)")
plt.ylim(0, 1.05)
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.savefig(os.path.join(output_dir, "Figure_7_Sensitivity_Coverage_vs_SINR_Threshold.png"), dpi=300)
plt.show()

display(sensitivity_df.round(3))

# ============================================================
# 11. Save Parameter Record
# ============================================================

params_text = f"""
Topic 5: Massive MIMO Antenna Panel Deactivation Simulation
Final Optimized Assignment-Aligned Version

Main Simulation Parameters:
- Cell radius: {cell_radius} m
- Total panels: {total_panels}
- Antennas per panel: {antennas_per_panel}
- Total antennas: {total_antennas}
- Bandwidth: {bandwidth_hz / 1e6} MHz
- Transmit power: {tx_power_dbm} dBm
- Noise power: {noise_power_dbm} dBm
- Main SINR threshold: {sinr_threshold_db} dB
- Monte Carlo trials: {num_trials}

Traffic Input:
- Synthetic 24-hour diurnal traffic profile generated using a Poisson process
- Representative validation states:
  - Low Load: {traffic_scenarios["Low Load"]} users
  - Medium Load: {traffic_scenarios["Medium Load"]} users
  - Peak Load: {traffic_scenarios["Peak Load"]} users

Strategies:
1. Baseline Always-On:
   - Low Load: 8 panels
   - Medium Load: 8 panels
   - Peak Load: 8 panels

2. Balanced Adaptive:
   - Low Load: 4 panels
   - Medium Load: 6 panels
   - Peak Load: 8 panels

3. Aggressive Adaptive:
   - Low Load: 2 panels
   - Medium Load: 4 panels
   - Peak Load: 8 panels

QoS Metrics:
- Average SINR
- Average throughput
- Coverage probability

Sensitivity Test:
- Coverage probability versus SINR threshold
- Thresholds tested: {sensitivity_thresholds_db} dB
- Scenario: {sensitivity_load}
"""

with open(os.path.join(output_dir, "simulation_parameters_final.txt"), "w") as f:
    f.write(params_text)

# ============================================================
# 12. Save README and requirements.txt
# ============================================================

readme_text = """
# Topic 5: Massive MIMO Antenna Panel Deactivation

## Project Description
This project evaluates traffic-aware antenna panel deactivation for an energy-efficient Massive MIMO base station. It compares an always-on baseline with balanced and aggressive adaptive strategies under low, medium, and peak traffic conditions.

## Assignment Alignment
The simulation includes:
- Component-based base station power model
- Massive MIMO antenna panel deactivation
- MRT beamforming
- Spatial user distribution with angular spread
- SINR, throughput, and coverage probability evaluation
- Synthetic diurnal traffic input
- Baseline versus proposed strategies across three traffic load scenarios
- Sensitivity testing of coverage probability versus SINR threshold

## Simulation Tool
Google Colab with Python.

## Required Libraries
- numpy
- pandas
- matplotlib

## How to Run
1. Open the notebook in Google Colab.
2. Run all cells from top to bottom.
3. The script automatically generates numerical outputs, figures, parameter files, and a ZIP archive.
4. Download the ZIP archive when execution is complete.

## Main Parameters
- Massive MIMO array: 64T64R
- Number of panels: 8
- Antennas per panel: 8
- Cell radius: 500 m
- Bandwidth: 20 MHz
- Transmit power: 43 dBm
- Noise power: -94 dBm
- Main SINR threshold: 2.4 dB
- Monte Carlo trials: 300

## Traffic Scenarios
- Low Load: 5 users
- Medium Load: 15 users
- Peak Load: 30 users

## Strategies
1. Baseline Always-On: 8 panels for all loads
2. Balanced Adaptive: 4 panels for low load, 6 panels for medium load, 8 panels for peak load
3. Aggressive Adaptive: 2 panels for low load, 4 panels for medium load, 8 panels for peak load

## Generated Outputs
- Figure_0_Synthetic_Diurnal_Traffic_Profile.png
- Figure_1_Power_Consumption.png
- Figure_2_Energy_Saving.png
- Figure_3_Average_SINR.png
- Figure_4_Average_Throughput.png
- Figure_5_Coverage_Probability.png
- Figure_6_Energy_Saving_vs_Coverage.png
- Figure_7_Sensitivity_Coverage_vs_SINR_Threshold.png
- simulation_results_final.csv
- sensitivity_results_coverage_vs_sinr_threshold.csv
- simulation_parameters_final.txt
- README.md
- requirements.txt
"""

with open(os.path.join(output_dir, "README.md"), "w") as f:
    f.write(readme_text)

with open(os.path.join(output_dir, "requirements.txt"), "w") as f:
    f.write("numpy\npandas\nmatplotlib\n")

# ============================================================
# 13. Zip and Download All Outputs
# ============================================================

zip_filename = "Topic5_Massive_MIMO_Final_Optimized_Outputs.zip"

with zipfile.ZipFile(zip_filename, "w") as zipf:
    for root, dirs, file_names in os.walk(output_dir):
        for file_name in file_names:
            file_path = os.path.join(root, file_name)
            zipf.write(file_path, arcname=file_name)

print("Final optimized simulation completed successfully.")
print(f"All files are saved in: {output_dir}")
print(f"Downloading: {zip_filename}")

files.download(zip_filename)
```
